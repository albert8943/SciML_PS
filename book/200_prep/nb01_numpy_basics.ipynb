Practice array creation, slicing, and a simple plot.


```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x) + 0.1 * np.random.randn(x.size)
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.show()
print(x.shape, float(y.mean()))
```
